```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.ensemble import IsolationForest
```

```python
# Generating normal data points (Clustered)
rng = np.random.RandomState(42)
X = 0.3 * rng.randn(100, 2)  # 100 normal points
```

```python
# Adding outliers (Far from normal cluster)
X_outliers = np.random.uniform(low=-4, high=4, size=(10, 2))  # 10 outliers
```

```python
# Combining normal and outlier data
X_all = np.vstack((X, X_outliers))
```

```python
# Convert to DataFrame for visualization
df = pd.DataFrame(X_all, columns=["Feature1", "Feature2"])
df["Anomaly"] = ["Normal"]*100 + ["Outlier"]*10  # Labeling for visualization
```

```python
# Train Isolation Forest Model
model = IsolationForest(contamination=0.1, random_state=42)
model.fit(X_all)
```

```python
# Predict Anomalies
df["Score"] = model.decision_function(X_all)  # Anomaly score
df["Anomaly_Pred"] = model.predict(X_all)  # -1 = anomaly, 1 = normal
```

```python
# Convert predictions to labels
df["Anomaly_Label"] = df["Anomaly_Pred"].map({1: "Normal", -1: "Outlier"})
```

```python
# Set color mapping
colors = {"Normal": "blue", "Outlier": "red"}

# Scatter Plot
plt.figure(figsize=(8,6))
sns.scatterplot(data=df, x="Feature1", y="Feature2", hue="Anomaly_Label", palette=colors)
plt.title("Isolation Forest Anomaly Detection")
plt.show()
```
